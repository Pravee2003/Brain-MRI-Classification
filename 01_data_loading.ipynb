```python
import os
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

# Change this path to your local path
DATASET_PATH = r"C:\Users\PRAVEEN S\Desktop\BrainTumor-Cloud-AI\data\raw\ASNR-MICCAI-BraTS2023-GLI-Challenge-TrainingData"

# Pick ONE patient folder
patient = os.listdir(DATASET_PATH)[0]
patient_path = os.path.join(DATASET_PATH, patient)

# Load FLAIR image
flair_path = [f for f in os.listdir(patient_path) if "t2w" in f][0]
flair_img = nib.load(os.path.join(patient_path, flair_path))
flair_data = flair_img.get_fdata()

print("FLAIR shape:", flair_data.shape)

# Visualize middle slice
slice_index = flair_data.shape[2] // 2
plt.imshow(flair_data[:, :, slice_index], cmap="gray")
plt.title("FLAIR MRI - Middle Slice")
plt.axis("off")
plt.show()
```
